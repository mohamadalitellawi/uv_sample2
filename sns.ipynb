```python
import pandas as pd
import numpy as np
import seaborn as sns
mortality_data = 'https://murach.com/python_analysis/mortality.csv'
```

```python
mortality_df = pd.read_csv(mortality_data)
mortality_df.head(4)
```

```python
sns.set_style('whitegrid')
g = sns.relplot(
    data=mortality_df,
    x='Year',
    y='Death Rate',
    hue='Age Group',
    kind='scatter',
    palette='bright',
    height=4,
    aspect=1.6,
)
```

```python
sns.relplot(
    data=mortality_df,
    x='Year',
    y='Death Rate',
    hue='Age Group',
    kind='scatter',
    palette='bright',
    height=4,
    aspect=1.6,
    col='Age Group',
    col_wrap=2,
    size='Death Rate',
    sizes=(10, 100),

)
```

```python
for ax in g.axes.flat:
    ax.set(
        title='Hello Mat',
        ylabel='Deaths per 100,000',
        xlabel='Year',
        xlim=(1990,2015),
        ylim=(0,150)
    )
g.figure
```

```python
g.savefig('xxx.pdf')
```

```python
g = sns.relplot(
    data=mortality_df,
    kind='line',
    x='Year',
    y='Death Rate',
    hue='Age Group',
)
```

```python
g = sns.catplot(
    data=mortality_df.query('Year in (2000,2010)'),
    kind='bar',
    x = 'Year',
    y = 'Death Rate',
    hue='Age Group',
    errorbar=None,
)
```
